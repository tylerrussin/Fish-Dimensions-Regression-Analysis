# Автоматический HTML-отчёт (дополнительная часть)

Генерация отчёта через `ydata-profiling` как дополнение к собственному EDA.

```python
from pathlib import Path
import warnings
import pandas as pd
from ydata_profiling import ProfileReport

warnings.filterwarnings("ignore", category=DeprecationWarning)

DATA_PATH = Path("../data/Fish.csv")
REPORT_PATH = Path("../reports/auto-report.html")

df = pd.read_csv(DATA_PATH)
print(f"Загружено: {df.shape[0]} строк × {df.shape[1]} колонок")
df.head()
```

```python
profile = ProfileReport(
    df,
    title="FishGrow — автоматический отчёт",
    explorative=True,
)

profile.to_file(REPORT_PATH)
print(f"Отчёт сохранён: {REPORT_PATH.resolve()}")
```

**Вывод.** Сгенерирован автоматический HTML-отчёт через `ydata-profiling`.
Он дополняет собственный EDA: содержит обзор всех признаков, распределения,
матрицу корреляций, карту пропусков, предупреждения о проблемах данных.

Открыть в браузере: `reports/auto-report.html`.
